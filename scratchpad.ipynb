```python
from datetime import datetime, timedelta

# Your input string
date_str = "26 sep 2026"

# 1. Clean "sept" to standard 3-letter "sep" for Python parsing
cleaned_str = date_str.replace("sept", "sep")

# 2. Parse the string into a datetime object
# %a = weekday, %d = day, %b = month, %Y = year
parsed_date = datetime.strptime(cleaned_str, "%d %b %Y")

# 3. Add 7 days
new_date = parsed_date + timedelta(days=7)

# 4. Format back to the original style string
output_str = new_date.strftime("%a %d %b %Y").lower()

# 5. Restore "sept" formatting if the new month is September
output_str = output_str.replace("sep", "sept")

print(output_str)  # Output: sat 03 oct 2026
```
